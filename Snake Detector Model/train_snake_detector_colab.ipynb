```python
# Step 3: Download the Snake Dataset
from roboflow import Roboflow

rf = Roboflow(api_key="nvSMsXnu4b62mWW7RTwG")
project = rf.workspace("aiexample-c8c2y").project("snake-emula")
version = project.version(2)
dataset = version.download("yolov8")
print("Dataset downloaded to:", dataset.location)
```

```python
# Step 4: Train YOLOv8 on GPU for 100 Epochs (~5-8 minutes total)
from ultralytics import YOLO

model = YOLO("yolov8n.pt")

results = model.train(
    data=f"{dataset.location}/data.yaml",
    epochs=100,
    imgsz=640,
    batch=16,
    device=0,
    name="snake_detector_colab"
)
```

```python
# Step 5: Automatically Download best.pt directly to your PC
from google.colab import files

best_weights = "runs/detect/snake_detector_colab/weights/best.pt"
files.download(best_weights)
print("Downloaded! Move this file into runs/detect/snake_detector_yolov8n/weights/")
```
